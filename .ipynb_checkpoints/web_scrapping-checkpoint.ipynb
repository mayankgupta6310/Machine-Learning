```python
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebkit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage = requests.get('https://www.ambitionbox.com/list-of-companies?page=336',headers= headers).text
```

```python
soup = BeautifulSoup(webpage, 'lxml')
```

```python
company = soup.find_all('div',class_="companyCardWrapper")
```

```python
len(company)
```

```python
name= []
rating = []
review =[]
for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find('div',class_="rating_star_container").text.strip())
    review.append(i.find('span',class_="companyCardWrapper__companyRatingCount").text.strip())

d= {'name':name,'rating':rating, 'review':review}
df = pd.DataFrame(d)
df
```
